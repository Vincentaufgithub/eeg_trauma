```python
def wavelet_scatter_transform(data, J, Q=8, T=None):
    if not T:
        T = len(data)

    # Compute scattering
    scattering = Scattering1D(J, T, Q)
    Sx = scattering(data) # values
    meta = scattering.meta() # meta information
    
    return Sx, meta
```

```python
Random_Seed_State = 91
J = 4 # Number of scales 2**4 -» 16 data points
Q = 2
chs = ['PO7',
       'PO3',
       'O1',
       'POz',
       'Oz',
       'PO4',
       'O2',
       'PO8',
       'Fp1',
       'FP2',
       'AF7',
       'AF3',
       'AFz',
       'AF4',
       'AF8',
       ]


data = []
labels = []
for condition in everything_epoched_dict.keys():
    condition_results = []
    
    for participant in everything_epoched_dict[condition]:

        condition_epochs = participant.get_data(picks=chs)
        
        for epoch in condition_epochs:
            epoch_results = []

            for channel in epoch:
                Sx, _ = wavelet_scatter_transform(channel, J=J, Q=Q)
                epoch_results.append(Sx.flatten(order="F")) # 0th, first and second order scattering coefficients all flattenedif len(channel) == 0:

            epoch_results_flattened = np.concatenate(epoch_results)
            condition_results.append(epoch_results_flattened)
        
    data.append(condition_results)


    


```

```python

train_ratio = .8
test_ratio = .2
Random_Seed_State = 91

n_trials = len(data[0]) #assuming we have the same n of trials for each condition. 
labels = np.array((np.ones(n_trials)*0,np.ones(n_trials)*1,np.ones(n_trials)*2,np.ones(n_trials)*3)).flatten().astype(int)
arrs_flattened = [arr for inner_list in wst_data for arr in inner_list] #flatten the data


x_train, x_test, y_train, y_test = train_test_split(arrs_flattened, labels, train_size = train_ratio, test_size = test_ratio, random_state = Random_Seed_State, stratify= labels )
print('Number of training data:', len(x_train),'  ', 'Number of test data:', len(x_test))
```
